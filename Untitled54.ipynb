La fonction resize_image redimensionne l'image si la taille depasse 500x500 


```python
import cv2

def resize_image(img, max_width=500, max_height=500):
   
    height, width = img.shape[:2]
    if width > max_width or height > max_height:
        scaling_factor = min(max_width / width, max_height / height)
        new_width = int(width * scaling_factor)
        new_height = int(height * scaling_factor)
        img = cv2.resize(img, (new_width, new_height), interpolation=cv2.INTER_AREA)
    return img
```

la fonction cartoonize_image permet de transformer nos images en style cartoon


```python


def cartoonize_image(image_path, resize=True):
    
    # Charger l'image
    img = cv2.imread(image_path)
    if img is None:
        raise ValueError(f"Impossible de charger l'image : {image_path}")

    # Redimensionner l'image si nécessaire
    if resize:
        img = resize_image(img)

    # Convertir en niveaux de gris
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    gray = cv2.medianBlur(gray, 7)

    # Détection des contours
    edges = cv2.adaptiveThreshold(gray, 255, cv2.ADAPTIVE_THRESH_MEAN_C, 
                                   cv2.THRESH_BINARY, 9, 9)

    # Ajustement des paramètres du filtre bilatéral en fonction de la taille de l'image
    height, width = img.shape[:2]
    d = max(5, min(15, int(min(height, width) / 100)))  # Diamètre dynamique
    sigmaColor = 100  # Ajustement fixe
    sigmaSpace = max(75, min(200, int(min(height, width) / 5)))  # Dynamique

    # Appliquer un filtre bilatéral pour lisser les couleurs
    color = cv2.bilateralFilter(img, d, sigmaColor, sigmaSpace)

    # Combiner les bords avec l'image colorée
    cartoon = cv2.bitwise_and(color, color, mask=edges)
    return cartoon


```

```python
import os
import cv2
from glob import glob
from sklearn.model_selection import train_test_split
import random



def process_and_split_dataset(input_folder, output_folder, test_size=0.2):
    """Traite les images et divise le jeu de données en ensembles d'entraînement et de test."""
    real_output = os.path.join(output_folder, "real")
    cartoon_output = os.path.join(output_folder, "cartoon")

    # Créer les dossiers de sortie
    os.makedirs(real_output, exist_ok=True)
    os.makedirs(cartoon_output, exist_ok=True)

    # Collecter toutes les images
    images = []
    for celebrity_folder in os.listdir(input_folder):
        celebrity_path = os.path.join(input_folder, celebrity_folder)
        if os.path.isdir(celebrity_path):
            images.extend(glob(f"{celebrity_path}/*.jpg"))

    # Diviser en ensembles d'entraînement et de test
    train_images, test_images = train_test_split(images, test_size=test_size, random_state=42)

    # Fonction pour traiter et sauvegarder les images
    def process_images(image_list, subset):
        for image_path in image_list:
            filename = os.path.basename(image_path)
            celebrity_name = os.path.basename(os.path.dirname(image_path))
            output_filename = f"{celebrity_name}_{filename}"

            # Lire l'image réelle
            real = cv2.imread(image_path)

            # Transformer en style cartoon
            cartoon = cartoonize_image(image_path)

            # Sauvegarder les résultats
            cv2.imwrite(os.path.join(real_output, subset + "_" + output_filename), real)
            cv2.imwrite(os.path.join(cartoon_output, subset + "_" + output_filename), cartoon)

    # Traiter les ensembles d'entraînement et de test
    process_images(train_images, "train")
    process_images(test_images, "test")

# Exemple d'utilisation
process_and_split_dataset("Celebrity_Faces", "dataset", test_size=0.2)
```

```python
from sklearn.model_selection import train_test_split
import shutil
import random 
def split_dataset(real_folder, cartoon_folder, output_folder, test_size=0.2):
    """
    Divise les données en ensembles d'entraînement et de test.
    """
    # Créer les dossiers de sortie
    train_real = os.path.join(output_folder, "train/real")
    train_cartoon = os.path.join(output_folder, "train/cartoon")
    test_real = os.path.join(output_folder, "test/real")
    test_cartoon = os.path.join(output_folder, "test/cartoon")

    os.makedirs(train_real, exist_ok=True)
    os.makedirs(train_cartoon, exist_ok=True)
    os.makedirs(test_real, exist_ok=True)
    os.makedirs(test_cartoon, exist_ok=True)

    # Collecter les fichiers
    real_images = sorted(glob(f"{real_folder}/*.jpg"))
    cartoon_images = sorted(glob(f"{cartoon_folder}/*.jpg"))

    # Vérifier que les deux ensembles correspondent
    assert len(real_images) == len(cartoon_images), "Le nombre d'images réelles et cartoonisées ne correspond pas."

    # Séparer les fichiers
    train_real_files, test_real_files, train_cartoon_files, test_cartoon_files = train_test_split(
        real_images, cartoon_images, test_size=test_size, random_state=42
    )

    # Copier les fichiers dans les dossiers appropriés
    for src, dest in zip(train_real_files, train_cartoon_files):
        shutil.copy(src, train_real)
        shutil.copy(dest, train_cartoon)

    for src, dest in zip(test_real_files, test_cartoon_files):
        shutil.copy(src, test_real)
        shutil.copy(dest, test_cartoon)

# Exemple d'utilisation
split_dataset("dataset/real", "dataset/cartoon", "dataset_split", test_size=0.2)
```

Préparation et sauvegarde du dataset pour l'entraînement et le test d'un modèle de conversion d'images en style cartoon

```python
from os import listdir
from numpy import asarray
from numpy import savez_compressed
from keras.preprocessing.image import load_img, img_to_array

# Charger toutes les images d'un répertoire en mémoire
def load_images(real_path, cartoon_path, size=(256, 256)):
    src_list, tar_list = list(), list()

    # Charger les images réelles
    for filename in listdir(real_path):
        # Charger et redimensionner l'image réelle
        real_pixels = load_img(real_path + filename, target_size=size)
        real_pixels = img_to_array(real_pixels)
        src_list.append(real_pixels)

        # Charger l'image correspondante en style cartoon
        if filename in listdir(cartoon_path):  # Vérifie si une image correspondante existe
            cartoon_pixels = load_img(cartoon_path + filename, target_size=size)
            cartoon_pixels = img_to_array(cartoon_pixels)
            tar_list.append(cartoon_pixels)

    # Convertir les listes en tableaux NumPy
    return [asarray(src_list), asarray(tar_list)]

# Chemin du dataset
train_real_path = 'dataset_split/train/real/'
train_cartoon_path = 'dataset_split/train/cartoon/'

# Charger les images d'entraînement
[src_images, tar_images] = load_images(train_real_path, train_cartoon_path)
print('Chargé: ', src_images.shape, tar_images.shape)

# Sauvegarder les données sous forme de fichier compressé
filename = 'train_data_256.npz'
savez_compressed(filename, src_images, tar_images)
print('Dataset sauvegardé : ', filename)

# Répéter pour les données de test
test_real_path = 'dataset_split/test/real/'
test_cartoon_path = 'dataset_split/test/cartoon/'

[test_src_images, test_tar_images] = load_images(test_real_path, test_cartoon_path)
print('Chargé (test): ', test_src_images.shape, test_tar_images.shape)

filename_test = 'test_data_256.npz'
savez_compressed(filename_test, test_src_images, test_tar_images)
print('Dataset de test sauvegardé : ', filename_test)
```

```python
import cv2
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D  # Nécessaire pour les visualisations 3D
import os
import numpy as np

# Chemins vers vos répertoires d'images
real_images_path = "dataset_split/train/real/"  # Répertoire des images réelles
cartoon_images_path = "dataset_split/train/cartoon/"  # Répertoire des images cartoonisées

# Obtenir la liste des fichiers d'images
real_images = sorted([os.path.join(real_images_path, img) for img in os.listdir(real_images_path) if img.endswith(('.png', '.jpg'))])
cartoon_images = sorted([os.path.join(cartoon_images_path, img) for img in os.listdir(cartoon_images_path) if img.endswith(('.png', '.jpg'))])

# Fonction pour charger et redimensionner une image
def load_and_resize(image_path, size=(256, 256)):
    image = cv2.imread(image_path)  # Charger l'image
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)  # Convertir de BGR (OpenCV) à RGB (matplotlib)
    image = cv2.resize(image, size)  # Redimensionner
    return image

# Fonction pour créer une visualisation 3D des caractéristiques spatiales
def visualize_spatial_features_3d(real_image, cartoon_image):
    # Convertir les images en niveaux de gris
    real_gray = cv2.cvtColor(real_image, cv2.COLOR_RGB2GRAY)
    cartoon_gray = cv2.cvtColor(cartoon_image, cv2.COLOR_RGB2GRAY)

    # Détection des contours (Edge Detection)
    real_edges = cv2.Canny(real_gray, 100, 200)
    cartoon_edges = cv2.Canny(cartoon_gray, 100, 200)

    # Créer une grille pour l'axe Z
    x = np.arange(real_gray.shape[1])
    y = np.arange(real_gray.shape[0])
    x, y = np.meshgrid(x, y)

    # Affichage en 3D
    fig = plt.figure(figsize=(20, 10))

    # Image Réelle (Contours)
    ax1 = fig.add_subplot(231, projection='3d')
    ax1.plot_surface(x, y, real_edges, cmap='viridis')
    ax1.set_title("Contours Réels (3D)")
    ax1.set_xlabel("X")
    ax1.set_ylabel("Y")
    ax1.set_zlabel("Intensité")

    # Image Cartoonisée (Contours)
    ax2 = fig.add_subplot(232, projection='3d')
    ax2.plot_surface(x, y, cartoon_edges, cmap='viridis')
    ax2.set_title("Contours Cartoon (3D)")
    ax2.set_xlabel("X")
    ax2.set_ylabel("Y")
    ax2.set_zlabel("Intensité")

    # Gradient Sobel X (Réel)
    real_sobelx = cv2.Sobel(real_gray, cv2.CV_64F, 1, 0, ksize=3)
    ax3 = fig.add_subplot(233, projection='3d')
    ax3.plot_surface(x, y, np.abs(real_sobelx), cmap='plasma')
    ax3.set_title("Gradient SobelX Réel (3D)")
    ax3.set_xlabel("X")
    ax3.set_ylabel("Y")
    ax3.set_zlabel("Intensité")

    # Gradient Sobel X (Cartoon)
    cartoon_sobelx = cv2.Sobel(cartoon_gray, cv2.CV_64F, 1, 0, ksize=3)
    ax4 = fig.add_subplot(234, projection='3d')
    ax4.plot_surface(x, y, np.abs(cartoon_sobelx), cmap='plasma')
    ax4.set_title("Gradient SobelX Cartoon (3D)")
    ax4.set_xlabel("X")
    ax4.set_ylabel("Y")
    ax4.set_zlabel("Intensité")

    # Image Réelle Originale
    ax5 = fig.add_subplot(235)
    ax5.imshow(real_image)
    ax5.set_title("Image Réelle")
    ax5.axis("off")

    # Image Cartoon Originale
    ax6 = fig.add_subplot(236)
    ax6.imshow(cartoon_image)
    ax6.set_title("Image Cartoonisée")
    ax6.axis("off")

    plt.tight_layout()
    plt.show()

# Boucle sur les premières images pour la visualisation
for real_path, cartoon_path in zip(real_images[:2], cartoon_images[:2]):  # Limité à 2 pour les tests
    real_image = load_and_resize(real_path)
    cartoon_image = load_and_resize(cartoon_path)
    visualize_spatial_features_3d(real_image, cartoon_image)
```

la fonction display_random_cartoon_images Affiche aléatoirement `num_images` images réelles et cartoonisées côte à côte, tout en respectant leur taille d'origine

```python
import os
import random
import cv2
import matplotlib.pyplot as plt

def display_random_cartoon_images(real_folder, cartoon_folder, num_images=10, target_size=(256, 256)):
    """
    Affiche des paires d'images réelles et cartoonisées redimensionnées à une taille standard.
    
    Args:
        real_folder (str): Dossier contenant les images réelles.
        cartoon_folder (str): Dossier contenant les images cartoonisées.
        num_images (int): Nombre d'images à afficher.
        target_size (tuple): Taille standard pour afficher les images (hauteur, largeur).
    """
    # Liste des fichiers disponibles dans les dossiers
    real_images = sorted(os.listdir(real_folder))
    cartoon_images = sorted(os.listdir(cartoon_folder))
    
    # Vérifier que les deux dossiers contiennent le même nombre d'images
    if len(real_images) != len(cartoon_images):
        raise ValueError("Le nombre d'images dans les dossiers `real` et `cartoon` ne correspond pas.")
    
    # Sélectionner aléatoirement des indices pour afficher des paires d'images
    indices = random.sample(range(len(real_images)), num_images)
    
    # Afficher les images
    for idx in indices:
        # Charger les images réelle et cartoonisée
        real_image_path = os.path.join(real_folder, real_images[idx])
        cartoon_image_path = os.path.join(cartoon_folder, cartoon_images[idx])
        
        real_image = cv2.imread(real_image_path)
        cartoon_image = cv2.imread(cartoon_image_path)
        
        # Redimensionner les images à la taille cible (target_size)
        real_image = cv2.resize(real_image, target_size)
        cartoon_image = cv2.resize(cartoon_image, target_size)
        
        # Convertir les images en RGB pour matplotlib
        real_image = cv2.cvtColor(real_image, cv2.COLOR_BGR2RGB)
        cartoon_image = cv2.cvtColor(cartoon_image, cv2.COLOR_BGR2RGB)
        
        # Afficher les images côte à côte
        plt.figure(figsize=(10, 5))
        
        # Afficher l'image réelle
        plt.subplot(1, 2, 1)
        plt.imshow(real_image)
        plt.title("Image Réelle")
        plt.axis("off")
        
        # Afficher l'image cartoonisée
        plt.subplot(1, 2, 2)
        plt.imshow(cartoon_image)
        plt.title("Image Cartoonisée")
        plt.axis("off")
        
        # Afficher les paires d'images
        plt.tight_layout()
        plt.show()

# Dossiers contenant les images
real_folder = "dataset_split/train/real"  # Remplacez par le chemin de vos images réelles
cartoon_folder = "dataset_split/train/cartoon"  # Remplacez par le chemin de vos images cartoonisées

# Afficher les images avec une taille standard de 256x256
display_random_cartoon_images(real_folder, cartoon_folder, num_images=10, target_size=(256, 256))
```

```python


import numpy as np
import tensorflow as tf
from numpy import savez_compressed

def preprocess_and_save(filename, output_file, resize_shape=(128, 128), max_samples=100):
    # Charger les données brutes
    data = np.load(filename)
    X1, X2 = data['arr_0'], data['arr_1']
    # Limiter le nombre d'échantillons
    X1, X2 = X1[:max_samples], X2[:max_samples]
    # Redimensionner
    X1 = np.array([tf.image.resize(img, resize_shape).numpy() for img in X1])
    X2 = np.array([tf.image.resize(img, resize_shape).numpy() for img in X2])
    # Sauvegarder les données prétraitées
    savez_compressed(output_file, arr_0=X1, arr_1=X2)
    print(f"Prétraitement terminé, sauvegardé dans {output_file}")

# Exécutez une fois pour prétraiter les données
preprocess_and_save('train_data_256.npz', 'preprocessed_data.npz')
```

2. **Entraîner le GAN Pix2Pix**
Chargement des images et définition du modèle discriminant pour un modèle GAN

```python
from os import listdir
from numpy import asarray, savez_compressed
from keras.preprocessing.image import load_img, img_to_array
from keras.models import Model
from keras.layers import Input, Concatenate, Conv2D, LeakyReLU, BatchNormalization, Activation
from keras.initializers import RandomNormal
from keras.optimizers import Adam

# Charger toutes les images d'un répertoire en mémoire
def load_images(real_path, cartoon_path, size=(256, 256)):
    src_list, tar_list = list(), list()

    # Charger les images réelles
    for filename in listdir(real_path):
        # Charger et redimensionner l'image réelle
        real_pixels = load_img(real_path + filename, target_size=size)
        real_pixels = img_to_array(real_pixels)
        src_list.append(real_pixels)

        # Charger l'image correspondante en style cartoon
        if filename in listdir(cartoon_path):  # Vérifie si une image correspondante existe
            cartoon_pixels = load_img(cartoon_path + filename, target_size=size)
            cartoon_pixels = img_to_array(cartoon_pixels)
            tar_list.append(cartoon_pixels)

    # Convertir les listes en tableaux NumPy
    return [asarray(src_list), asarray(tar_list)]


# Définir le modèle du discriminateur

def define_discriminator(image_shape):
    init = RandomNormal(stddev=0.02)
    # Entrée de l'image réelle
    in_src_image = Input(shape=image_shape)
    # Entrée de l'image générée (cible)
    in_target_image = Input(shape=image_shape)
    # Fusion des deux images sur les canaux
    merged = Concatenate()([in_src_image, in_target_image])
    # Bloc C64
    d = Conv2D(64, (4, 4), strides=(2, 2), padding='same', kernel_initializer=init)(merged)
    d = LeakyReLU(alpha=0.2)(d)
    # Bloc C128
    d = Conv2D(128, (4, 4), strides=(2, 2), padding='same', kernel_initializer=init)(d)
    d = BatchNormalization()(d)
    d = LeakyReLU(alpha=0.2)(d)
    # Bloc C256
    d = Conv2D(256, (4, 4), strides=(2, 2), padding='same', kernel_initializer=init)(d)
    d = BatchNormalization()(d)
    d = LeakyReLU(alpha=0.2)(d)
    # Bloc C512
    d = Conv2D(512, (4, 4), strides=(2, 2), padding='same', kernel_initializer=init)(d)
    d = BatchNormalization()(d)
    d = LeakyReLU(alpha=0.2)(d)
    # Couche finale (patches)
    d = Conv2D(1, (4, 4), padding='same', kernel_initializer=init)(d)
    patch_out = Activation('sigmoid')(d)
    # Modèle final
    model = Model([in_src_image, in_target_image], patch_out)
    # Compilation
    opt = Adam(learning_rate=0.0002, beta_1=0.5)
    model.compile(loss='binary_crossentropy', optimizer=opt, loss_weights=[0.5])
    return model
```

Définition d'un modèle générateur U-Net pour la conversion d'images avec blocs d'encodage et de décodage

```python
from keras.models import Model
from keras.layers import Input, Conv2D, Conv2DTranspose, BatchNormalization, Dropout, Concatenate, LeakyReLU, Activation, UpSampling2D
from keras.initializers import RandomNormal

# Définir un bloc d'encodage
def define_encoder_block(layer_in, n_filters, batchnorm=True):
    # Initialisation des poids
    init = RandomNormal(stddev=0.02)
    # Couche de réduction
    g = Conv2D(n_filters, (4, 4), strides=(2, 2), padding='same', kernel_initializer=init)(layer_in)
    # Ajouter Batch Normalization (si activé)
    if batchnorm:
        g = BatchNormalization()(g)
    # Activation LeakyReLU
    g = LeakyReLU(alpha=0.2)(g)
    return g

# Définir un bloc de décodage


def decoder_block(input_tensor, skip_tensor, n_filters):
    # Sur-échantillonnage de l'entrée
    g = Conv2DTranspose(n_filters, (4, 4), strides=(2, 2), padding='same')(input_tensor)
    g = BatchNormalization()(g)
    g = Dropout(0.5)(g) if n_filters > 64 else g
    # Ajuster la taille du skip_tensor si nécessaire
    if g.shape[1] != skip_tensor.shape[1] or g.shape[2] != skip_tensor.shape[2]:
        skip_tensor = Conv2D(n_filters, (1, 1), padding='same')(skip_tensor)
    # Concaténation
    g = Concatenate()([g, skip_tensor])
    g = Activation('relu')(g)
    return g



# Définir le modèle générateur U-Net
def define_generator(image_shape):
    init = 'he_normal'
    # Encodeur
    in_image = Input(shape=image_shape)
    e1 = Conv2D(64, (4, 4), strides=(2, 2), padding='same', kernel_initializer=init)(in_image)
    e1 = LeakyReLU(alpha=0.2)(e1)
    e2 = Conv2D(128, (4, 4), strides=(2, 2), padding='same', kernel_initializer=init)(e1)
    e2 = BatchNormalization()(e2)
    e2 = LeakyReLU(alpha=0.2)(e2)
    e3 = Conv2D(256, (4, 4), strides=(2, 2), padding='same', kernel_initializer=init)(e2)
    e3 = BatchNormalization()(e3)
    e3 = LeakyReLU(alpha=0.2)(e3)
    e4 = Conv2D(512, (4, 4), strides=(2, 2), padding='same', kernel_initializer=init)(e3)
    e4 = BatchNormalization()(e4)
    e4 = LeakyReLU(alpha=0.2)(e4)

    # Bottleneck
    b = Conv2D(512, (4, 4), strides=(2, 2), padding='same', kernel_initializer=init)(e4)
    b = Activation('relu')(b)

    # Décodeur
    d1 = decoder_block(b, e4, 512)
    d2 = decoder_block(d1, e3, 256)
    d3 = decoder_block(d2, e2, 128)
    d4 = decoder_block(d3, e1, 64)

    # Sortie
    g = Conv2DTranspose(3, (4, 4), strides=(2, 2), padding='same', kernel_initializer=init)(d4)
    out_image = Activation('tanh')(g)

    # Modèle
    model = Model(in_image, out_image)
    return model

```

Définition du Modèle GAN combiné pour l'entraînement Pix2Pix

```python
def define_gan(g_model, d_model, image_shape):
    # Congeler les poids du discriminateur pendant l'entraînement du générateur
    for layer in d_model.layers:
        if not isinstance(layer, BatchNormalization):
            layer.trainable = False
    # Entrée
    in_src = Input(shape=image_shape)
    # Sortie générée
    gen_out = g_model(in_src)
    # Discriminateur connecté
    dis_out = d_model([in_src, gen_out])
    # Modèle combiné
    model = Model(in_src, [dis_out, gen_out])
    # Compilation
    opt = Adam(learning_rate=0.0002, beta_1=0.5)
    model.compile(loss=['binary_crossentropy', 'mae'], optimizer=opt, loss_weights=[1, 100])
    return model
```

Chargement des Données et Entraînement d'un Modèle Pix2Pix GAN


```python
from os import listdir
from numpy import asarray, load, zeros, ones
from numpy.random import randint
from keras.models import Model
from keras.layers import Input, Conv2D, Conv2DTranspose, LeakyReLU, Activation, Concatenate, Dropout, BatchNormalization
from keras.optimizers import Adam
from matplotlib import pyplot
import numpy as np

# Charger les données compressées et normalisées
def load_real_samples(filename):
    # Charger les tableaux compressés
    data = load(filename)
    X1, X2 = data['arr_0'], data['arr_1']
    # Normaliser les pixels dans l'intervalle [-1, 1]
    X1 = (X1 - 127.5) / 127.5
    X2 = (X2 - 127.5) / 127.5
    return [X1, X2]

# Charger les données préparées
dataset = load_real_samples('preprocessed_data.npz')
print('Données chargées :', dataset[0].shape, dataset[1].shape)

# Générer un batch d'exemples réels
def generate_real_samples(dataset, n_samples, patch_shape):
    trainA, trainB = dataset
    ix = randint(0, trainA.shape[0], n_samples)
    X1, X2 = trainA[ix], trainB[ix]
    y = ones((n_samples, patch_shape, patch_shape, 1))
    return [X1, X2], y

# Générer un batch d'exemples fictifs
def generate_fake_samples(g_model, samples, patch_shape):
    X = g_model.predict(samples)
    y = zeros((len(X), patch_shape, patch_shape, 1))
    return X, y

# Sauvegarder les résultats intermédiaires et le modèle
def summarize_performance(step, g_model, dataset, n_samples=3):
    [X_realA, X_realB], _ = generate_real_samples(dataset, n_samples, 1)
    X_fakeB, _ = generate_fake_samples(g_model, X_realA, 1)
    # Re-normalisation des pixels pour affichage
    X_realA = (X_realA + 1) / 2.0
    X_realB = (X_realB + 1) / 2.0
    X_fakeB = (X_fakeB + 1) / 2.0

    # Afficher et sauvegarder les images
    for i in range(n_samples):
        pyplot.subplot(3, n_samples, 1 + i)
        pyplot.axis('off')
        pyplot.imshow(X_realA[i])
    for i in range(n_samples):
        pyplot.subplot(3, n_samples, 1 + n_samples + i)
        pyplot.axis('off')
        pyplot.imshow(X_fakeB[i])
    for i in range(n_samples):
        pyplot.subplot(3, n_samples, 1 + n_samples*2 + i)
        pyplot.axis('off')
        pyplot.imshow(X_realB[i])

    filename1 = f'plot_{step+1:06d}.png'
    pyplot.savefig(filename1)
    pyplot.close()
    filename2 = f'model_{step+1:06d}.h5'
    g_model.save(filename2)
    print(f'>Saved: {filename1} and {filename2}')

# Fonction d'entraînement du Pix2Pix GAN
# Fonction d'entraînement du Pix2Pix GAN
def train(d_model, g_model, gan_model, dataset, n_epochs=30, n_batch=8):
    # Taille des patches de sortie du discriminateur
    n_patch = d_model.output_shape[1]
    trainA, trainB = dataset
    bat_per_epo = int(len(trainA) / n_batch)

    for epoch in range(n_epochs):
        print(f"Époque {epoch + 1}/{n_epochs}")
        for batch in range(bat_per_epo):
            # Générer des exemples réels
            [X_realA, X_realB], y_real = generate_real_samples(dataset, n_batch, n_patch)
            # Générer des exemples fictifs
            X_fakeB, y_fake = generate_fake_samples(g_model, X_realA, n_patch)
            # Mettre à jour le discriminateur
            d_loss1 = d_model.train_on_batch([X_realA, X_realB], y_real)
            d_loss2 = d_model.train_on_batch([X_realA, X_fakeB], y_fake)
            # Mettre à jour le générateur
            g_loss = gan_model.train_on_batch(X_realA, [y_real, X_realB])
            # Afficher les pertes
            print(f'>Batch {batch+1}/{bat_per_epo}, d1[{d_loss1:.3f}] d2[{d_loss2:.3f}] g[{g_loss[0]:.3f}]')

    # Sauvegarder les performances et le modèle à la dernière époque
    print(f"Sauvegarde du modèle et des résultats à l'époque finale ({n_epochs})...")
    summarize_performance(n_epochs - 1, g_model, dataset)


# Définir les formes d'entrée
image_shape = dataset[0].shape[1:]

# Définir les modèles
# Assurez-vous d'avoir défini `define_discriminator`, `define_generator`, et `define_gan` précédemment
d_model = define_discriminator(image_shape)
g_model = define_generator(image_shape)
gan_model = define_gan(g_model, d_model, image_shape)

# Entraîner le modèle Pix2Pix GAN
train(d_model, g_model, gan_model, dataset)
```

```python

import numpy as np
import cv2
import matplotlib.pyplot as plt
from keras.models import load_model

def load_and_preprocess_image(image_path, target_size):
    """
    Charge et prétraite une image pour le modèle.
    Args:
        image_path (str): Chemin de l'image.
        target_size (tuple): Taille cible (hauteur, largeur) pour redimensionner l'image.
    Returns:
        np.array: Image prétraitée prête pour le modèle.
    """
    # Charger l'image avec OpenCV
    image = cv2.imread(image_path)
    if image is None:
        raise ValueError(f"Impossible de charger l'image : {image_path}")
    # Redimensionner l'image à la taille cible
    image = cv2.resize(image, target_size)
    # Convertir l'image en RGB
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    # Normaliser l'image dans l'intervalle [-1, 1]
    image = (image - 127.5) / 127.5
    # Ajouter une dimension batch
    image = np.expand_dims(image, axis=0)
    return image

def display_results(input_image, output_image):
    """
    Affiche l'image d'entrée et la sortie générée côte à côte.
    Args:
        input_image (np.array): Image d'entrée.
        output_image (np.array): Image générée par le modèle.
    """
    # Re-normaliser les images pour l'affichage (intervalle [0, 1])
    input_image = (input_image + 1) / 2.0
    output_image = (output_image + 1) / 2.0

    # Supprimer la dimension batch
    input_image = input_image[0]
    output_image = output_image[0]

    # Afficher les résultats
    plt.figure(figsize=(10, 5))
    plt.subplot(1, 2, 1)
    plt.axis("off")
    plt.title("Image Réelle")
    plt.imshow(input_image)
    plt.subplot(1, 2, 2)
    plt.axis("off")
    plt.title("Image Générée")
    plt.imshow(output_image)
    plt.tight_layout()
    plt.show()

def test_model_on_real_image(model_path, image_path):
    """
    Teste un modèle sur une image réelle.
    Args:
        model_path (str): Chemin du fichier du modèle enregistré (format `.h5`).
        image_path (str): Chemin de l'image réelle.
    """
    # Charger le modèle
    model = load_model(model_path)
    print(f"Modèle chargé depuis : {model_path}")
    
    # Déterminer la taille d'entrée attendue par le modèle
    target_size = model.input_shape[1:3]  # (hauteur, largeur)
    
    # Charger et prétraiter l'image réelle
    input_image = load_and_preprocess_image(image_path, target_size)
    print(f"Image chargée et prétraitée depuis : {image_path}")
    
    # Générer une prédiction avec le modèle
    output_image = model.predict(input_image)
    print("Prédiction générée.")
    
    # Afficher les résultats
    display_results(input_image, output_image)

# Exemple d'utilisation
model_path = "model_000030.h5"  # Remplacez par le chemin de votre modèle
image_path = "test_Sandra Bullock_075_4b6e8283.jpg"  # Remplacez par le chemin de votre image réelle
test_model_on_real_image(model_path, image_path)
```

Afficher une démo sur une vidéo(video de Messi)

```python


import cv2


def cartoonize_video(video_path, model_path, output_path):
    # Charger le modèle
    model = load_model(model_path)

    # Charger la vidéo d'entrée
    cap = cv2.VideoCapture(video_path)
    fourcc = cv2.VideoWriter_fourcc(*'XVID')
    out = cv2.VideoWriter(output_path, fourcc, 20.0, (128, 128))

    while cap.isOpened():
        ret, frame = cap.read()
        if not ret:
            break

        # Prétraitement de l'image
        frame_resized = cv2.resize(frame, (128, 128))
        input_image = (frame_resized - 127.5) / 127.5
        input_image = np.expand_dims(input_image, axis=0)

        # Prédiction
        output_image = model.predict(input_image)
        output_image = ((output_image[0] + 1) / 2.0 * 255).astype(np.uint8)

        # Ajouter au fichier vidéo
        out.write(cv2.cvtColor(output_image, cv2.COLOR_RGB2BGR))

        # Afficher la frame cartoonisée
        cv2.imshow("Frame Cartoonisée", cv2.cvtColor(output_image, cv2.COLOR_RGB2BGR))

        # Arrêter le traitement si 'q' est pressé
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    # Libérer les ressources
    cap.release()
    out.release()
    cv2.destroyAllWindows()
    print("Vidéo cartoonisée enregistrée et affichée.")

# Exemple d'utilisation
cartoonize_video("Messi.mp4", "model_000030.h5", "output_video.avi")

```
